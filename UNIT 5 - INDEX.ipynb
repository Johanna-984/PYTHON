```python
# The pivot_table Method

food = pd.read_csv("foods.csv")

food.head()
```

```python
# Media

food.pivot_table(values="Spend", index="Gender")
```

```python
# agregar una funcion
food.pivot_table(values="Spend", index="Gender", aggfunc="mean")
```

```python
# sumar valores
food.pivot_table(values="Spend", index="Gender", aggfunc="sum")
```

```python
# valores de gasos para cada articulo
# cuanto se ha vendido por producto

food.pivot_table(values="Spend", index="Item", aggfunc="sum")
```

```python
# por grupo y por articulo: suma acumulada de los gastos para cada combinacion de sexo y articulo

food.pivot_table(values="Spend", index=["Gender","Item"], aggfunc="sum")
```

```python
food.pivot_table(values="Spend", index=["Gender","Item"], columns="City", aggfunc="sum")
```

```python
food.pivot_table(values="Spend", index="Item", columns=["Gender", "City"], aggfunc="sum")
```

```python
food.pivot_table(values="Spend", index="Item", columns=["Gender", "City"], aggfunc="count")
```

```python
food.pivot_table(values="Spend", index="Item", columns=["Gender", "City"], aggfunc="max")
```

```python
food.pivot_table(values="Spend", index="Item", columns=["Gender", "City"], aggfunc="min")
```

```python
food.pivot_table(values="Spend", index="Item", columns=["Gender", "City"], aggfunc="std")
```
