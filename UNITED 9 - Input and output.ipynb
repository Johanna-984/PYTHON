```python
import requests

url = 'https://data.cityofnewyork.us/api/views/25th-nujf/rows.csv'

try:
    response = requests.get(url, verify=False)
    response.raise_for_status()  # Lanza una excepción para códigos de estado HTTP malos
    print("Descarga exitosa")
    # Puedes guardar el contenido del CSV en un archivo
    with open('datos_nyc.csv', 'wb') as f:
        f.write(response.content)
    print("Datos guardados en datos_nyc.csv")
except requests.exceptions.RequestException as e:
    print(f"Ocurrió un error: {e}")
```
