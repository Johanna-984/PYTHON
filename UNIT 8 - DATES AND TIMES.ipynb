```python
# En resumen, stocks.index - pd.DateOffset(days=5) es una herramienta flexible para manipular y 
# analizar datos de series de tiempo, permitiéndote crear referencias temporales pasadas para 
# comparaciones, generar rangos de fechas desplazados y realizar alineaciones basadas en desfases de tiempo. 
# La aplicación específica dependerá del contexto de tu análisis
stocks.index + pd.DateOffset(days=5)
stocks.index - pd.DateOffset(days=5)
stocks.index - pd.DateOffset(months=5)
stocks.index - pd.DateOffset(years=1)
stocks.index - pd.DateOffset(hours=7)
stocks.index + pd.DateOffset(years =1, months=3, days=2, hours=14, minutes=23, seconds=12)
```

```python
birthdays = pd.date_range(start="2000-12-12", end="2023-04-12", freq=pd.DateOffset(years=1))
birthdays
```

```python
stocks[stocks.index.isin(birthdays)]
```

```python
fecha_referencia = pd.to_datetime('2014-03-10')
indice_desplazado = stocks.index - pd.DateOffset(days=5)
# Esto devolverá True para las fechas en stocks.index que, al restar 5 días, son iguales a la fecha de referencia
mascara = indice_desplazado == fecha_referencia
datos_desplazados = stocks[mascara]
```

```python
serie_A = pd.Series(range(len(stocks)), index=stocks.index)
serie_B_alineada = pd.Series(range(len(stocks)), index=stocks.index - pd.DateOffset(days=5))

# Ahora podrías intentar alinear serie_A con los valores de serie_B que ocurrieron 5 días antes
df_comparacion = pd.DataFrame({'A': serie_A, 'B_retrasada': serie_B_alineada}).reindex(stocks.index)
df_comparacion
```

```python
dates = pd.read_csv("ecommerce.csv", parse_dates=["order_date", "delivery_date"]).set_index("ID")

dates
```

```python
dates["Delivery Time"] = dates["delivery_date"] - dates["order_date"]
```

```python
dates.head()
```

```python
# fechas futuras

dates["If it took twice as long"] = dates["delivery_date"] + dates["Delivery Time"]

dates.head()
```
