```python
# Filter with more than One condition (AND)
employees = pd.read_csv('employees.csv', parse_dates=['Start Date'], date_format="%m/%d/%Y")
employees['Last Login Time'] = pd.to_datetime(employees["Last Login Time"], format="%H:%M %p").dt.time
employees['Senior Management'] = employees['Senior Management'].astype(bool)
employees['Gender'] = employees['Gender'].astype('category')
employees.head()
```

```python
# Mujeres que trabajan en Marketing
is_female = employees['Gender'] == 'Female'
is_in_marketing = employees['Team'] == 'Marketing'
salary_over_100k = employees['Salary'] > 100000
```

```python
is_female & is_in_marketing & salary_over_100k
# True, True -> True
#'''Si es mujer pero no de marketing'''
#True, False -> False
#'''Si no es mujer pero si de marketing'''
#False, True -> False
#'''No es mujer pero tampoco de marketing'''
# False, False -> False
```

```python
employees[is_female & is_in_marketing & salary_over_100k]
```

```python
#Filter with more than One Condition
# AND vs OR
#
# True, Tru -> True
# True, False -> True
# False, True -> True
# False, False -> False

#Employees who are either senior managment OR started before January 1, 1990

is_senior_management = employees["Senior Management"]
started_in_80s = employees['Start Date'] < "1990-01-01"

employees[is_senior_management | started_in_80s]
```

```python
#The isin method
#Legal Team or Sales Team or Product Team
# Se debe cumplir por lo menos una de estas tres condiciones

# Filter with more than One condition (AND)
import pandas as pd
employees = pd.read_csv('employees.csv', parse_dates=['Start Date'], date_format="%m/%d/%Y")
employees['Last Login Time'] = pd.to_datetime(employees["Last Login Time"], format="%H:%M %p").dt.time
employees['Senior Management'] = employees['Senior Management'].astype(bool)
employees['Gender'] = employees['Gender'].astype('category')
employees.head()

legal = employees['Team'] == 'Legal'
sales = employees['Team'] == 'Sales'
product = employees['Team'] == 'Product'

employees[legal | sales | product]
```

```python
# Comprueba si este valor esta dentro de las listas
# Verifica si cada elemento de una serie o DF esta presente en un a lista o en otra serie. 
# Devuelve una serie booleana que inidca si (True) esta presente en la lista e indica si no lo esta (False)
employees[employees["Team"].isin(["Legal", "Sales", "Product"])]
```

```python
# The isnull and notnull Methods
employees = pd.read_csv('employees.csv', parse_dates=['Start Date'], date_format="%m/%d/%Y")
employees['Last Login Time'] = pd.to_datetime(employees["Last Login Time"], format="%H:%M %p").dt.time
employees['Senior Management'] = employees['Senior Management'].astype(bool)
employees['Gender'] = employees['Gender'].astype('category')
employees.head()
```

```python
# isnull va a marcar los valores NaN o los valores que faltan con true

# isnull se utiliza para determinar si un valor en un objeto de pandas es nulo o faltante, true = valor nulo, false = si no lo es
# notnull()  se utiliza para determinar si un calor en un objeto Pandas no es nulo ni faltante, true = si es lavlor no es nulo, false = si lo es

employees[employees['Team'].isnull()]
employees[employees['Team'].notnull()]

# Marco de datos filtrado donde se va a tener un valor nulo en la columna de nombre y valor presente en la columna equipo 
# solo se han seleccionado los empleados cuyo nombre falta pero tienen un equipo asignado.
employees[employees['First Name'].isnull() & employees['Team'].notnull()]
```

```python
# The between method

# Se usa para filtrar filas de DF dentro de un iintervalo determinado

# quiero extraer a los empleados con un salario entre 130,000 

employees[employees['Salary'].between(60000,70000)]
```

```python
employees[employees['Bonus %'].between(2.0,5.0)]
```

```python
# filtro between con tiempo

employees[employees['Start Date'].between('1991-01-01','1992-01-01')]
```

```python
import datetime as dt

# dt.time(8,30) de la manana
# dt.time(12,0) del mediodia


employees[employees['Last Login Time'].between(dt.time(8,30), dt.time(12,0))]
```

```python
# The duplicated Method: identificar las filas duplicadas en un DF mediante el metodo duplicado
# todas las filas de employees donde el valor en la columna 'First Name' aparece más de una vez, excluyendo la primera vez que aparece cada nombre duplicado.
employees[employees['First Name'].duplicated()]
employees[employees['First Name'].duplicated(keep='first')]
employees[employees['First Name'].duplicated(keep='last')]
# ~ revierte los booleanos
# significa que los eventos ocurren una vez en el conjunto de datos original.
employees[~employees['First Name'].duplicated(keep=False)]
```
